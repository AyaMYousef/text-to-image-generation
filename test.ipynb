```python
from langchain_ollama import ChatOllama
from agno.agent import Agent
from agno.models.ollama import Ollama




llm = ChatOllama(
    model="llama3.2:latest",
    temperature=0.7,)
```

```python
llm.invoke("""
Your task is to provide information and explanation about the next prompt
Always the answer in 3 lines

PROMPT :
An X-ray of a human chest showing pneumonia in the right lowe lobe
""").content
```

'This X-ray image shows a medical condition called pneumonia, which is an infection in one or both lungs. The right lower lobe of the lung appears inflamed and filled with fluid, indicating a bacterial or viral infection. The white patches or nodules on the lung tissue are also indicative of pneumonia, a common respiratory illness.'

```python
from agno.memory.v2.db.sqlite import SqliteMemoryDb
from agno.memory.v2.memory import Memory

memory_db = SqliteMemoryDb(table_name="memory", db_file="tmp/memory.db")
memory = Memory(db=memory_db)
```

```python
john_doe_id = "john_doe@example.com"

agent = Agent(
    model=Ollama(id="llama3.2:latest"),
    instructions=["you will be given a prompt and you have to provide information and explanation about the prompt in 3 lines",
                  "If the prompt contains a disease name, you have to provide information about the disease",
                  "your task is to be like a guide for medical enthusiasts",],
    # num_history_responses=3,
    memory=memory,
    # enable_agentic_memory=True,
    add_history_to_messages=True
)
```

```python
output = agent.run("""
describe it for me
""",
user_id=john_doe_id)
```

```python
output.content
```

"**Pneumonia Overview**\n\nPneumonia is an infection that inflames the air sacs in one or both lungs. It can be caused by bacteria, viruses, or fungi and typically affects people with weakened immune systems or those who are elderly. The x-ray image you provided shows signs of pneumonia in the right lower lobe, indicating an infection that has spread to this part of the lung.\n\n**X-Ray Findings**\n\nThe x-ray image reveals several characteristics of pneumonia:\n\n* **Infiltrates**: White patches on the x-ray indicate areas of consolidation, where fluid and pus have accumulated due to inflammation.\n* **Increased Vascular Marking**: Blood vessels are more visible on the x-ray, indicating increased blood flow to the affected area.\n* **Lobe-Specific Involvement**: The pneumonia is confined to the right lower lobe, which is a common site for this type of infection.\n\n**Treatment and Management**\n\nPneumonia typically requires antibiotics or antiviral medications to treat the underlying infection. In severe cases, hospitalization may be necessary to provide oxygen therapy, fluids, and rest. It's essential to seek medical attention immediately if you suspect pneumonia symptoms, such as difficulty breathing, fever, or chest pain.\n\n**Complications and Prognosis**\n\nIf left untreated or not managed properly, pneumonia can lead to serious complications, including respiratory failure, sepsis, or even death. However, with prompt medical attention and treatment, most people recover fully from pneumonia."
